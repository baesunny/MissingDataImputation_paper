```python
import pandas as pd

file_path = '../data/Imputation_Test.xlsx'
sheet_index = 0
data = pd.read_excel(file_path, sheet_name=sheet_index)
```

```python
data.head(10)
```

```python
data.info()
```

---

```python
df = data.copy()
df = df.iloc[:, 3:]
```

```python
df.info()
```

```python
stocking = df.iloc[:, :13]
stocking.columns = stocking.iloc[1]
stocking = stocking.drop([0, 1]).reset_index(drop=True)
stocking
```

```python
middle = df.iloc[:, 13:16]
middle.columns = middle.iloc[1]
middle = middle.drop([0, 1]).reset_index(drop=True)
middle
```

```python
ctp = df.iloc[:, 16:53]
ctp.columns = ctp.iloc[1]
ctp = ctp.drop([0, 1]).reset_index(drop=True)
ctp
```

```python
ctq = df.iloc[:, 53:66]
ctq.columns = ctq.iloc[1]
ctq = ctq.drop([0, 1]).reset_index(drop=True)
ctq
```

---

```python
ctp.info()
```

```python
temp = ctp.iloc[:, 4:]
temp = temp.apply(lambda col: pd.to_numeric(col, errors='coerce'))
temp = temp.fillna(0)
temp
```

```python
ctp = pd.concat([ctp.iloc[:, :4], temp], axis=1)
ctp
```

```python
ctp.info()
```

---

```python
stocking.to_csv("stocking.csv", index=True)
middle.to_csv("middle.csv", index=True)
ctp.to_csv("ctp.csv", index=True)
ctq.to_csv("ctq.csv", index=True)
```
